# Evidence before generation
This independent consumer uses its bundled fictional fixture. No producer, API token or paid LLM is required. Install the project and run from the notebooks directory.
Learning goals: contract verification, fixed analytical tools, prediction provenance and the boundary between software tests and actual model evaluation.

```python
from pathlib import Path
from arena_copilot.snapshot import default_fixture, validate_snapshot
from arena_copilot.tools import Tools
from arena_copilot.agent import Copilot
snapshot = default_fixture()
manifest = validate_snapshot(snapshot)
manifest['data_kind'], manifest['dataset_version'], manifest['as_of']
```

```python
tools = Tools(snapshot)
tools.call('get_team_stats', {'sport': 'valorant', 'team_name': 'Neon Wolves'})
```

## A language model explains a forecast; it does not create its probabilities
Read an exact precomputed distribution, data kind and feature cutoff. Interpret the evidence without presenting an uncalibrated synthetic model as a real prediction service.

```python
predictions = tools.read_json('predictions.json')
prediction = predictions[0]
assert abs(sum(prediction['probabilities'].values()) - 1) < 1e-6
tools.call('get_prediction', {'match_id': prediction['match_id']})
```

## Deterministic evidence browser
This explicit mode is useful for software checks. It is never counted as generative inference or silently substituted for an unavailable model.

```python
agent = Copilot(snapshot, Path('../data/notebook-01'))
answer = agent.ask('Show VALORANT team win rates', mode='evidence')
answer['citations'], answer['metrics']
```
